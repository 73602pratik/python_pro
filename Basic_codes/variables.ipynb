A customer enters

- Account Balance
- Withdrawal Amount

Rules

-If sufficient balance → deduct amount.
Otherwise print "Insufficient Funds".
Also display remaining balance after a successful transaction.

Industry

- Banking
- ATM software
- Payment gateways



```python
account_balance = float(input('Enter account balance: '))
withdrawal_amount = float(input('Enter withdrawal amount: '))

if account_balance >= withdrawal_amount:
    account_balance -= withdrawal_amount
    print(f"Withdrawal successful. New balance: {account_balance}")
    print(f"Withdrawal Amount: {withdrawal_amount}")
else:
    print("Insufficient balance for withdrawal.")
```

## Question 2: Cloud Server Storage Monitor

A DevOps engineer wants to check server usage.

Input

- Total Storage (GB)
- Used Storage (GB)

Calculate

- Available Storage
- Used Percentage

If usage exceeds 90%

Print

- Storage Alert

Otherwise

- Storage Healthy

Industry

- AWS
- Azure
- Google Cloud

```python
total_storage  = float(input('Enter total storage in GB: '))
used_storage = float(input('Enter used storage in GB: '))
available_storage = total_storage - used_storage
used_percentage = (used_storage / total_storage) * 100
print(f"Available Storage: {available_storage} GB")
print(f"Used Storage Percentage: {used_percentage}%")
if used_percentage > 90:
    print("Warning: Storage usage exceeds 90%! Consider freeing up space.")
else:
    print("Storage usage is within acceptable limits.")
```

## Question 3: E-commerce Order Summary

Customer enters

- Product Name
- Quantity
- Unit Price

Calculate

- Total
- Discount (5% if total > ₹5000)
- GST (18%)
- Final Payable Amount

Print a professional invoice.

Industry

- Amazon

- Flipkart

- Myntra

```python
product_name = input('Enter product name: ')
product_quantity = int(input('Enter product quantity: '))
unit_price = float(input('Enter unit price: '))
total_cost = product_quantity * unit_price

if total_cost > 5000:
    discount = total_cost * 0.05
else:
    discount = 0

gst = (total_cost - discount) * 0.18
final = total_cost - discount + gst

print(f"Product: {product_name}")
print(f"Quantity: {product_quantity}")
print(f"Unit Price: {unit_price}")
print(f"Total Cost: {total_cost}")
print(f"Discount: {discount}")
print(f"GST: {gst}")
print(f"Final Amount: {final}")
```

## ERP Billing Systems

* Challenge Question (Bonus)
Company Expense Tracker

- A startup tracks monthly expenses.

Take

-- Office Rent
-- Internet Bill
-- Electricity Bill
-- Employee Salary
-- Miscellaneous Expenses

Calculate

- Total Expense
- Monthly Budget (₹5,00,000)
- Remaining Budget

If expenses exceed budget

- Budget Exceeded

Else

- Budget Remaining

```python
office_rent = float(input("Enter Office Rent: ₹"))
internet_bill = float(input("Enter Internet Bill: ₹"))
electricity_bill = float(input("Enter Electricity Bill: ₹"))
employee_salary = float(input("Enter Employee Salary: ₹"))
misc_expense = float(input("Enter Miscellaneous Expense: ₹"))

monthly_budget = 500000
```

```python
expenses = [
    office_rent,
    internet_bill,
    electricity_bill,
    employee_salary,
    misc_expense
]

total_expense = sum(expenses)
remaining_budget = monthly_budget - total_expense
```

```python
print("\n========== COMPANY EXPENSE REPORT ==========")
print(f"Office Rent          : ₹{office_rent}")
print(f"Internet Bill        : ₹{internet_bill}")
print(f"Electricity Bill     : ₹{electricity_bill}")
print(f"Employee Salary      : ₹{employee_salary}")
print(f"Miscellaneous        : ₹{misc_expense}")
print("--------------------------------------------")
print(f"Total Expense        : ₹{total_expense}")
print(f"Monthly Budget       : ₹{monthly_budget}")
print(f"Remaining Budget     : ₹{remaining_budget}")

if total_expense > monthly_budget:
    print("\nStatus : Budget Exceeded")
else:
    print("\nStatus : Budget Remaining")
```
