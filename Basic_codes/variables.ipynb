#### What is variable = identifier that reference objects and everything in python is an object

### - One of the biggest misconceptions beginners have is: Variables storr values.
### -> A python variable stores a reference to an object, not the object itself.

#### Why does this exist?
- price ------> [100]
- price = 100 # not this 
- It's imp when objects are mutable

#### when should we use it?
* always - Every lines of python code creates or manipulates objects.
##### When should we not think about it?
* for every small scripts you don't need to consciously think about object references.

```python
# simple example
age = 25
print(age)  # age --> object(25)
```

```python
# simple example
a = [1,2,3]
b=a
b.append(4)
print(a) # a --> object([1,2,3,4])
print(b) # b --> object([1,2,3,4])
# both variables a and b point to the same object in memory, so changes made through one variable are reflected in the other.
```

```python
a = [1,2]
b = a
c = a.copy() # creates a new object in memory
b.append(3)

print(a) # a --> object([1,2,3])
print(b) # b --> object([1,2,3])
print(c) # c --> object([1,2])
# shallow vs deep copy topic 
```

Company Expense Tracker

- A startup tracks monthly expenses.

Take

-- Office Rent
-- Internet Bill

-- Employee Salary
-- Miscellaneous Expenses

Calculate

- Total Expense
- Monthly Budget (₹5,00,000)
- Remaining Budget

If expenses exceed budget

Else

```python
office_rent = float(input("Enter Office Rent: ₹"))
internet_bill = float(input("Enter Internet Bill: ₹"))
electricity_bill = float(input("Enter Electricity Bill: ₹"))
employee_salary = float(input("Enter Employee Salary: ₹"))
misc_expense = float(input("Enter Miscellaneous Expense: ₹"))

monthly_budget = 500000
```

```python
expenses = [
    office_rent,
    internet_bill,
    electricity_bill,
    employee_salary,
    misc_expense
]

total_expense = sum(expenses)
remaining_budget = monthly_budget - total_expense
```

```python
print("\n========== COMPANY EXPENSE REPORT ==========")
print(f"Office Rent          : ₹{office_rent}")
print(f"Internet Bill        : ₹{internet_bill}")
print(f"Electricity Bill     : ₹{electricity_bill}")
print(f"Employee Salary      : ₹{employee_salary}")
print(f"Miscellaneous        : ₹{misc_expense}")
print("--------------------------------------------")
print(f"Total Expense        : ₹{total_expense}")
print(f"Monthly Budget       : ₹{monthly_budget}")
print(f"Remaining Budget     : ₹{remaining_budget}")

if total_expense > monthly_budget:
    print("\nStatus : Budget Exceeded")
else:
    print("\nStatus : Budget Remaining")
```

Mini Project
Project: Inventory Management CLI (Part 1)

Objective: Build the foundation of a command-line inventory system.

Features:

- Add products
- View products
- Update stock
- Store product data in memory using dictionaries and lists
- Use meaningful variable names and avoid accidental shared references

```python
inventory = []

def display_menu():
    print("\n=========================================")
    print("          INVENTORY MANAGEMENT           ")
    print("=========================================")
    print("1. Add Product")
    print("2. View Products")
    print("3. Update Stock")
    print("4. Exit")

def add_product():
    prod_id = input("\nProduct ID: ")
    prod_name = input("Product Name: ")
    prod_price = float(input("Product Price: "))
    prod_stock = int(input("Product Stock: "))
    prod_category = input("Product Category: ")

    product = {
        "id": prod_id,
        "name": prod_name,
        "price": prod_price,
        "stock": prod_stock,
        "category": prod_category,
    }

    inventory.append(product)
    print("\nProduct added successfully!")

def view_products():
    if not inventory:
        print("\nNo products in inventory.")
        return
    print("-----------------------------------------")
    print(f"{'ID':<10}{'Name':<20}{'Price':<10}{'Stock':<10}{'Category':<15}")
    print("-----------------------------------------")

    for item in inventory:
        print(f"{item['id']:<10}{item['name']:<20}{item['price']:<10}{item['stock']:<10}{item['category']:<15}")

def update_stock():
    prod_id = input("\nEnter Product ID to update stock: ")
    for item in inventory:
        if item['id'] == prod_id:
            new_stock = int(input("Enter new stock quantity: "))
            item['stock'] = new_stock
            print("\nStock updated successfully!")
            return
    print("\nProduct not found.")

# Main program loop
while True:
    display_menu()
    choice = input("Enter your choice (1-4): ")

    if choice == '1':
        add_product()
    elif choice == '2':
        view_products()
    elif choice == '3':
        update_stock()
    elif choice == '4':
        print("\nExiting Inventory Management. Goodbye!")
        break
    else:
        print("\nInvalid choice! Please select a valid option.")
```

```python
## == vs is
```

```python
## == checks for value equality, meaning it checks if the values of two objects are the same.
## is checks for identity equality, meaning it checks if two variables point to the same object in memory.
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
print(a == b)  # True, because the values are the same
print(a is b)  # False, because they are different objects in memory
# if value is None:
```

```python
## immutable
## int, float, str, tuple, bool, frozenset, bytes, complex, range, NoneType are immutable types in Python.

## mutable
## list, dict, set are mutable types in Python.
```

```python
## challenge
a = [1,2,3]
b = a
c = [1,2,3] 

print(a == b)  # True, because they reference the same object
print(a is b)  # True, because they reference the same object

print(a == c)  # True, because the values are the same
print(a is c)  # False, because they are different objects in memory
```
